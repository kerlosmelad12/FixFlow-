```python
import requests

payload = {
    "query": """
<p>I want to revert a particular commit in git.  Unfortunately, our organization still uses CVS as a standard, so when I commit back to CVS multiple git commits are rolled into one.  In this case I would love to single out the original git commit, but that is impossible.</p>

"""
}

resp = requests.post(
    "http://127.0.0.1:5000/Fixflow-V1/data/upload/",
    json=payload
)

print(resp.status_code)
print(resp.json())
```
