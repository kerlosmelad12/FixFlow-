```python
import requests

payload = {
    "query": """
<p>I have a database where one table &quot;Common&quot; is referenced by several other tables.  I wanted to see what records in the Common table were orphaned (i.e., had no references from any of the other tables).</p>
<p>I ran this query:</p>
<pre><code>select *
from Common
where common_id not in (select common_id from Table1)
and common_id not in (select common_id from Table2)
</code></pre>
<p>I know that there are orphaned records, but no records were returned.  Why not?</p>

'"""
}

resp = requests.post(
    "http://127.0.0.1:5000/Fixflow-V1/data/upload/",
    json=payload
)

print(resp.status_code)
print(resp.json())
```

```python
import requests

url = "https://api.stackexchange.com/2.3/search/advanced"

params = {
    "site": "stackoverflow",
    "q": "check if a string contains a string in c++",
    "sort": "relevance",
    "pagesize": 10,
    "filter": "withbody"
}

response = requests.get(url, params=params)

print(response.status_code)

data = response.json()

for item in data.get("items", []):
    print("=" * 80)
    print("TITLE:", item["title"])
    print("QUESTION ID:", item["question_id"])
    print("BODY:")
    print(item.get("body"))
    print("URL:", item["link"])
```

```python
import requests

question_id = 2340281

url = f"https://api.stackexchange.com/2.3/questions/{question_id}/answers"

params = {
    "site": "stackoverflow",
    "sort": "votes",
    "order": "desc",
    "filter": "withbody"
}

response = requests.get(url, params=params)

print(response.status_code)

data = response.json()

for answer in data.get("items", []):
    print("=" * 80)
    print("ANSWER ID:", answer["answer_id"])
    print("SCORE:", answer["score"])
    print("ACCEPTED:", answer.get("is_accepted", False))
    print("BODY:")
    print(answer["body"])
```
