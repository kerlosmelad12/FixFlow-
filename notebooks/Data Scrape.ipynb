```python
import requests
import random
import time
import json

url = "https://api.stackexchange.com/2.3/questions"
allquestions = []

pages = random.sample(range(1, 400), 30)

for page in pages:
    params = {
        "order": "desc",
        "sort": "votes ",
        "site": "stackoverflow",
        "pagesize": 100,
        "filter": "withbody",
        "page": page,
        "key": "rl_ox3yetMypPB4GUx8evpzAsiVi"
    }

    try:
        response = requests.get(url, params=params, timeout=10)

        if response.status_code != 200:
            print(f"Error: {response.status_code}")
            continue

        if not response.text.strip():
            print("Empty response")
            continue

        data = response.json()

    except Exception as e:
        print("Request failed:", e)
        continue

    filtered = [q for q in data.get('items', []) if q['answer_count'] > 5]
    allquestions.extend(filtered)

    print(f"Page {page} done, total: {len(allquestions)}")

    time.sleep(1.5)  

print("Final:", len(allquestions))

import json

with open("question.json", 'w', encoding='utf-8') as f:
    json.dump(allquestions, f, ensure_ascii=False, indent=4)
```

```python
import requests
import time
import pandas as pd
questions=pd.read_json('question.json')
questionsids=questions['question_id'].values
allanswers = []
count=0
answers_url_template = "https://api.stackexchange.com/2.3/questions/{ids}/answers" 
ans_params = { "order": "desc", "sort": "votes", "site": "stackoverflow", "filter": "withbody", "key": "rl_ox3yetMypPB4GUx8evpzAsiVi" }
for id in questionsids:
    url = answers_url_template.format(ids=id)

    try:
        response = requests.get(url, params=ans_params, timeout=5)

        if response.status_code != 200:
            print(f"Error {response.status_code} for question {id}")
            continue

        if not response.text.strip():
            print(f"Empty response for {id}")
            continue

        data = response.json()

    except Exception as e:
        print(f"Failed for {id}: {e}")
        continue

    if "backoff" in data:
        time.sleep(data["backoff"])

    items = data.get("items", [])
    allanswers.extend(items)
    count+=1

    print(f"Number of Question {count} Question {id} done, total answers: {len(allanswers)}")

    time.sleep(1.5)  
```

```python
import json
with open("answers.json", 'w', encoding='utf-8') as f:
    json.dump(allanswers, f, ensure_ascii=False, indent=4)
```
